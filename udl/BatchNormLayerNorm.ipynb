```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Define the grid size for the stacked cubes
x = 3
y = 8
z = 8

# Create the figure with a larger size
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Define the size of each cube
cube_size = 1

# Set offsets to space out the cubes along the x-axis
offsets = [(0, 0, 0), (x * 2, 0, 0), (x * 4, 0, 0), (x * 6, 0, 0)]

# Plot four separate sets of stacked cubes in a horizontal line
for n in range(4):
    offset_x, offset_y, offset_z = offsets[n]  
    for i in range(x):
        for j in range(y):
            for k in range(z):
                # First column along the x-axis is orange
                color = 'orange' if i == 0 else 'lightgray'
                
                ax.bar3d(i + offset_x, j + offset_y, k + offset_z, cube_size, cube_size, cube_size, color=color, shade=True)

# Set axis limits for better visualization
ax.set_xlim([0, x * 8])  
ax.set_ylim([0, y])  
ax.set_zlim([0, z])  

# Adjust the aspect ratio of the plot
ax.set_box_aspect([x * 8, y, z])  

# Set axis labels with improved readability
ax.set_xlabel('Batch=4  Channel=3  BatchNorm', fontsize=12, labelpad=15)
ax.set_ylabel('H = 8', fontsize=12, labelpad=-15)
ax.set_zlabel('W = 8', fontsize=12, labelpad=-15)

# Remove ticks
ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])

# Adjust view angle
ax.view_init(elev=25, azim=135)

# Optimize layout
plt.tight_layout()
plt.show()

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Define the grid size for the stacked cubes
x = 3
y = 8
z = 8

# Create the figure with a larger size
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Define the size of each cube
cube_size = 1

# Set offsets to space out the cubes along the y-axis (instead of x-axis)
offsets = [(0, 0, 0), (0, y + 2, 0), (0, (y + 2) * 2, 0), (0, (y + 2) * 3, 0)]

# Plot four separate sets of stacked cubes along the y-axis
for n in range(4):
    offset_x, offset_y, offset_z = offsets[n]  
    for i in range(x):
        for j in range(y):
            for k in range(z):
                # First column along the x-axis is orange
                color = 'orange' if i == 0 else 'lightgray'
                
                ax.bar3d(i + offset_x, j + offset_y, k + offset_z, cube_size, cube_size, cube_size, color=color, shade=True)

# Set axis limits for better visualization
ax.set_xlim([0, x])  
ax.set_ylim([0, (y + 2) * 4])  # Adjusted y limit
ax.set_zlim([0, z])  

# Adjust the aspect ratio of the plot
ax.set_box_aspect([x, (y + 2) * 4, z])  

# Set axis labels
ax.set_xlabel('Channel=3', fontsize=12, labelpad=-15)
ax.set_ylabel('                         Batch=4                            H = 8', fontsize=12, labelpad=10)
ax.set_zlabel('W = 8', fontsize=12, labelpad=-10)

# Remove ticks
ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])

# Adjust view angle
ax.view_init(elev=25, azim=135)

# Optimize layout
plt.tight_layout()
plt.show()


import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

# Define the grid size for the stacked cubes
x = 3
y = 8
z = 8

# Create the figure with a larger size
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Define the size of each cube
cube_size = 1

# Set offsets to space out the cubes along the y-axis (instead of x-axis)
offsets = [(0, 0, 0), (0, y + 2, 0), (0, (y + 2) * 2, 0), (0, (y + 2) * 3, 0)]

# Plot four separate sets of stacked cubes along the y-axis
for n in range(4):
    offset_x, offset_y, offset_z = offsets[n]  
    for i in range(x):
        for j in range(y):
            for k in range(z):
                # Set color: First cube (n == 0) is yellow, others are light gray
                color = 'orange' if n == 0 else 'lightgray'
                
                ax.bar3d(i + offset_x, j + offset_y, k + offset_z, cube_size, cube_size, cube_size, color=color, shade=True)

# Set axis limits for better visualization
ax.set_xlim([0, x])  
ax.set_ylim([0, (y + 2) * 4])  # Adjusted y limit
ax.set_zlim([0, z])  

# Adjust the aspect ratio of the plot
ax.set_box_aspect([x, (y + 2) * 4, z])  

# Set axis labels
# Set axis labels
ax.set_xlabel('Channel=3', fontsize=12, labelpad=-15)
ax.set_ylabel('                         Batch=4                            H = 8', fontsize=12, labelpad=10)
ax.set_zlabel('W = 8', fontsize=12, labelpad=-10)

# Remove ticks
ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])

# Adjust view angle
ax.view_init(elev=25, azim=135)

# Optimize layout
plt.tight_layout()
plt.show()

```
