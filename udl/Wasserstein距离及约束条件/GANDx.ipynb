```python
import numpy as np
import matplotlib.pyplot as plt

# 定义真实数据和生成数据的分布
def p_data(x):
    return 0.8 * np.exp(-0.5 * ((x - 1) / 0.5)**2)  # N(1, 0.5^2)

def p_g(x):
    return 0.6 * np.exp(-0.5 * ((x + 1) / 0.8)**2)  # N(-1, 0.8^2)

# 最优判别器 D*(x)
def D_star(x):
    return p_data(x) / (p_data(x) + p_g(x))

# 绘制 D(x), D(x*) 和 D*(x)
x_vals = np.linspace(-3, 3, 500)
real_dist = p_data(x_vals)
gen_dist = p_g(x_vals)
d_star_vals = D_star(x_vals)

plt.plot(x_vals, real_dist, label="Real Data $p_{data}(x)$", color="blue")
plt.plot(x_vals, gen_dist, label="Generated Data $p_g(x)$", color="red")
plt.plot(x_vals, d_star_vals, label="Optimal Discriminator $D^*(x)$", linestyle="--", color="green")
plt.xlabel("x")
plt.ylabel("Probability Density / Optimal D(x)")
plt.legend()
plt.title("Optimal Discriminator Matching Posterior Probability")
plt.show()
```
